```python
import os
import glob
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
import kagglehub

# 1. Download dataset using kagglehub
path = kagglehub.dataset_download("oddrationale/mnist-in-csv")
print("Path to dataset files:", path)

# Locate the train and test CSV files dynamically
csv_files = glob.glob(os.path.join(path, "*.csv"))
train_path = next(f for f in csv_files if 'train' in f.lower())
test_path = next(f for f in csv_files if 'test' in f.lower())

# Load datasets into DataFrames
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print(f"\nTrain set shape: {train_df.shape}")
print(f"Test set shape : {test_df.shape}")

# Separate labels and feature matrices
y_train = train_df.iloc[:, 0].values
X_train = train_df.iloc[:, 1:].values

y_test = test_df.iloc[:, 0].values
X_test = test_df.iloc[:, 1:].values

# Reshape features into 28x28x1 images and normalize pixel values (0 to 1)
X_train = X_train.reshape(-1, 28, 28, 1).astype('float32') / 255.0
X_test = X_test.reshape(-1, 28, 28, 1).astype('float32') / 255.0

# 2. Build CNN architecture using Keras
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 3. Display model summary
print("\n--- Model Summary ---")
model.summary()

# 4. Train the model for 5 epochs
print("\n--- Training Model ---")
model.fit(X_train, y_train, epochs=5, batch_size=64, validation_split=0.1)

# 5. Evaluate the model using the testing dataset
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\n--- Model Evaluation ---")
print(f"Test Accuracy: {test_acc:.4f}")

# 6. Predict the class of one test image
sample_img = np.expand_dims(X_test[0], axis=0)
prediction = model.predict(sample_img, verbose=0)
predicted_class = np.argmax(prediction)

print(f"\n--- Single Sample Prediction ---")
print(f"Actual Label   : {y_test[0]}")
print(f"Predicted Class: {predicted_class}")
```
