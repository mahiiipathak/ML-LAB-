```python
import os
import glob
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import kagglehub

# 1. Download dataset using kagglehub and load into Pandas DataFrame
path = kagglehub.dataset_download("rkiattisak/salaly-prediction-for-beginer")
print("Path to dataset files:", path)

# Locate the CSV file inside the downloaded directory dynamically
csv_files = glob.glob(os.path.join(path, "*.csv"))
if not csv_files:
    raise FileNotFoundError("No CSV file found in the downloaded dataset directory.")

dataset_path = csv_files[0]
df = pd.read_csv(dataset_path)

# 2. Display the first five records of the dataset
print("\n--- 1. First 5 Records of Original Dataset ---")
print(df.head())

# 3. Identify all missing values in the dataset
print("\n--- 2. Missing Value Summary ---")
print(df.isnull().sum())

# Separate numerical and categorical columns
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(include=['object']).columns.tolist()

# 4. Replace missing values in numerical columns using Mean Imputation
if num_cols:
    num_imputer = SimpleImputer(strategy='mean')
    df[num_cols] = num_imputer.fit_transform(df[num_cols])

# 5. Replace missing values in categorical columns using Most Frequent (Mode) value
if cat_cols:
    cat_imputer = SimpleImputer(strategy='most_frequent')
    df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])

print("\n--- 3. Dataset After Mean & Mode Imputation ---")
print(df.head())

# 6. Convert categorical attributes into numerical form using One-Hot Encoding
if cat_cols:
    df_encoded = pd.get_dummies(df, columns=cat_cols, drop_first=True)
else:
    df_encoded = df.copy()

print("\n--- 4. Encoded Feature Columns ---")
print(df_encoded.head())

# 7. Standardize all numerical features using StandardScaler
scaler = StandardScaler()
df_encoded[num_cols] = scaler.fit_transform(df_encoded[num_cols])

# 8. Display the transformed dataset and its dimensions after preprocessing
print("\n--- 5. Final Transformed Dataset ---")
print(df_encoded.head())
print("\nFinal Transformed Dataset Dimensions:", df_encoded.shape)
```
